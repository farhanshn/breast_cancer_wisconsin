خب وارد مرحله اخر شدیم  اول کتابخانه ها را ایمپورت می کنیم  و سپس از الگریتم اس وی ام استفاده می کنیم  طبق تحقیقات  برای زمانی که دیتا ها زیاد هستند مناسب است 

```python
# Import libraries required for model training

from pathlib import Path
import sys

import pandas as pd
import numpy as np

# Import machine learning models
from sklearn.svm import SVC


# Import model evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    f1_score,
    classification_report
)

# Import visualization library
import matplotlib.pyplot as plt
```

project path

```python
# Define the project root directory

PROJECT_ROOT = Path.cwd().parent

# Add the project root to Python's module search path

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

#define processed data directory

```python
# Define the directory containing processed datasets

PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"
```

laod processed data

حالا اینجا چهار  فایل سی اس وی که ساخته بودیم رو صدا میزنیم و لود می کنیم 

```python
# Load processed training and testing datasets

X_train = pd.read_csv(
    PROCESSED_DATA_DIR / "X_train.csv"
)

X_test = pd.read_csv(
    PROCESSED_DATA_DIR / "X_test.csv"
)

y_train = pd.read_csv(
    PROCESSED_DATA_DIR / "y_train.csv"
).squeeze()

y_test = pd.read_csv(
    PROCESSED_DATA_DIR / "y_test.csv"
).squeeze()
```

در اینجا استاد توضیح میدهد که میتوانیم چهار فایل دیتا ست رو به یک دیتا فریم جامع تبدیل کنیم اما در اینجا ضروری نیست  

حالا وقتشه شکل داده ها رو چک کنیم 

```python
# Verify the shape of the loaded datasets

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

در سلول پایین می خوایم تعداد واقعی هر کلاس رو مشخص کنیم

```python
# Check the target distribution in training and testing datasets

print("Training target distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting target distribution:")
print(y_test.value_counts().sort_index())
```

creatr svm

```python
# Create the SVM classification model

svm_model = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    probability=True,
    random_state=42
)
```

train model

```python
# Train the SVM model using the training data

svm_model.fit(
    X_train,
    y_train
)
```

make prediction 

```python
# Generate predictions for the test dataset

y_pred = svm_model.predict(X_test)
y_pred
```

در سلول پایین ده نمونه اول را کنار هم میگذاریم و چک می کنیم 


```python
# Display the first ten actual and predicted values

comparison = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': y_pred[:10]
})

comparison
```

check accuracy

```python
# Calculate the accuracy of the SVM model

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Accuracy: {accuracy:.2f}")
```

دقت مدل واقعا ایده ال است  در اینجا استاد عمق الگریتم ومدل رو میسنجد اما  مدل اس وی ام واقعا در عددی ایده ال است اما 
جهت بهتر بودن پروژه ان را میسنجیم 
خب بریم برای سنجش میزان پیچیدگی 

```python
# Check the complexity of the SVM model

print("Support vectors per class:", svm_model.n_support_)
print("Total support vectors:", svm_model.n_support_.sum())
```

support vector 

ساپورت ویکتور روشی برای سنجش میزان پیچیدگی در اس وی هست 

حالا در سلول بالا فهمیدیم میزان پیچیدگی مدل و تعداد ساپورت وکتور ها رو فهمیدیم 

```python
### Checking out a few different C values

c_values = [0.01, 0.1, 1, 10, 100, 1000]

for c in c_values:
    model = SVC(
        C=c,
        kernel="rbf",
        gamma="scale",
        random_state=42
    )

    model.fit(X_train, y_train)

    train_acc = model.score(X_train, y_train)
    test_acc = model.score(X_test, y_test)

    print(
        f'C:{c} | '
        f'Train acc: {train_acc:.3f} | '
        f'Test acc: {test_acc:.3f} |'
    )
```

خب بریم برای یک تحلیل کوچک در سلول بالا 
استاد در ویدیو خود عمق های متفاوت درخت تصمیم گیری را با برسی دقت تست و دقت ترین مورد مقایسه قرار داد
حالا ما هم شروع به مقایسه دقت ترین و تست می کنیم

خب حالا بریم برای تحلیل نتایج 
در اولین سی که یک صدم هست  دقت تست و ترین برابر شصت و دو و شصت سه هست 
که یعنی اندر فیتینگ

در دومین سی یعنی یک دهم دقت تست و ترین حدود نود چهار و نود پنج است که  به نسبت بهتر هستند 

در سومین و چهارمین تست و ترین هر دو در دقت ایده ال قرار دارند  پس ما یک تا ده را ایده ال میدانیم 

صد تا هزار در واقع اور فیت میشن و برای مدل ما مناسب نیستند 

خب حالا بریم برای ریبیلدینگ کردن 

```python
# Rebuild the SVM model with the selected C value

best_c = 1

final_model = SVC(
    C=best_c,
    kernel="rbf",
    gamma="scale",
    probability=True,
    random_state=42
)

final_model.fit(X_train, y_train)

y_pred = final_model.predict(X_test)
y_pred
```

خب من برای ریبیلدینگ سی یک رو انتخاب کردم با دقت تست و ترین نود و هفت و نود و هشت  در سلول بالا میبینید

```python
# Calculate the confusion matrix

cm = confusion_matrix(y_test, y_pred)
cm
```

CHECK PRECISION AND RECALL


```python
# Calculate precision and recall

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print('Precision: ', precision)
print('Recall: ', recall)
```

```python
# Calculate the F1-score

f1 = f1_score(y_test, y_pred)
print("F1-Score:", f1)
```
